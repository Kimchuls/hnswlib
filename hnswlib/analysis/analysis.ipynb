```python
import matplotlib.pyplot as plt

# Style configuration
style_cfg = {
    "title_size": 30,
    "label_size": 30,
    "legend_size": 32,
    "tick_size": 28,
    "marker_size": 5,
    "bar_hatches": ['/', '\\', '|', '-', 'x', '+', 'o', 'O', '.', '*'],
    "markers": ['o', 's', '^', 'd', '*', 'X', 'P', 'v', 'p'],
    "colors": ['darkblue', 'darkorange', 'green', 'red', 'darkred',
               'mediumorchid', 'saddlebrown', 'indigo', 'darkolivegreen']
}

# Read memory log files
def read_mem_log(filename):
    times = []
    mems = []
    with open(filename, 'r') as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith('#'):
                continue
            t, m = line.split()
            times.append(float(t))
            mems.append(float(m)/1024)
    return times, mems

# Load data
t1, mem1 = read_mem_log('mem_sift10m_me.log')
t2, mem2 = read_mem_log('mem_sift10m_m2.log')

# Plot
plt.figure(figsize=(12, 8))

# Plot lines with custom colors and markers
plt.plot(t1, mem1,
         label='Memory-Efficient',
         color=style_cfg['colors'][0],
         marker=style_cfg['markers'][0],
         markersize=style_cfg['marker_size'],
         linewidth=5)

plt.plot(t2, mem2,
         label='All-in-Memory',
         color=style_cfg['colors'][1],
         marker=style_cfg['markers'][1],
         markersize=style_cfg['marker_size'],
         linewidth=5)

# Labels, title and legend with custom sizes
plt.xlabel('Time (s)', fontsize=style_cfg['label_size'])
plt.ylabel('Memory (GB)', fontsize=style_cfg['label_size'])
# plt.title('Memory Usage Over Time', fontsize=style_cfg['title_size'])
plt.ylim(0, 25)
plt.xlim(0,600)

# Tick parameters
plt.tick_params(axis='both', which='major', labelsize=style_cfg['tick_size'])
plt.tick_params(axis='both', which='minor', labelsize=style_cfg['tick_size'])

# Legend in upper right with custom font size
# plt.legend(loc='upper right', fontsize=style_cfg['legend_size'])
plt.legend(
    loc='upper center',            # 放在上方中央
    bbox_to_anchor=(0.44, 1.2),     # (x,y) 坐标：x=0.5 居中，y>1 则在顶框之上
    ncol=2,                         # 两列
    fontsize=style_cfg['legend_size'],
    frameon=False                   # 可根据需要去掉边框
)

plt.tight_layout()
plt.savefig("memory-efficiency-comparison.png")
plt.show()
```
